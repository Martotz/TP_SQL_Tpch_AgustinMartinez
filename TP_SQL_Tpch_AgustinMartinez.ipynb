### Ejercicio 1.3 — Contar y resumir

#### Calculamos cantidad total de clientes, saldo promedio, máximo y mínimo.

```python
%sql
-- Calcular cantidad total de clientes y resumen de saldos
SELECT 
  COUNT(*) AS total_clientes,
  AVG(c_acctbal) AS saldo_promedio,
  MAX(c_acctbal) AS saldo_maximo,
  MIN(c_acctbal) AS saldo_minimo
FROM samples.tpch.customer;
```

### Ejercicio 1.4 — Columnas calculadas

##### Creamos una columna `tipo_cliente` que clasifica según el saldo: 'VIP' si supera 5000, 'estándar' si no.

```python
%sql
-- Clasificar clientes según su saldo
SELECT 
  c_name AS nombre_cliente,
  c_acctbal AS saldo_cuenta,
  CASE 
    WHEN c_acctbal > 5000 THEN 'VIP'
    ELSE 'estándar'
  END AS tipo_cliente
FROM samples.tpch.customer
ORDER BY saldo_cuenta DESC;
```

##### Como solo vemos "VIPs", verificamos si trae los datos de salario maximos y minimos

```python
%sql
SELECT 
  MIN(c_acctbal) AS saldo_minimo,
  MAX(c_acctbal) AS saldo_maximo
FROM samples.tpch.customer;
```

##### Buscamos los clientes "estandar"

```python
%sql
SELECT 
  c_name AS nombre_cliente,
  c_acctbal AS saldo_cuenta,
  CASE 
    WHEN c_acctbal > 5000 THEN 'VIP'
    ELSE 'estándar'
  END AS tipo_cliente
FROM samples.tpch.customer
ORDER BY saldo_cuenta ASC;
```

> Nota: El rango de saldos va de -999.99 a 9999.99, por lo que existen clientes con saldo negativo (deudores). 
> Al ordenar por saldo descendente, solo se ven los clientes con saldo alto (VIP). 
> Para visualizar los 'estándar', se puede ordenar ascendentemente o ajustar el umbral.

# Parte 2 - Filtros y agrupaciones 

La tabla orders registra cada pedido. La tabla lineitem detalla cada línea de producto
dentro de un pedido.

![image_1780267883797.png](./image_1780267883797.png "image_1780267883797.png")

En esta sección aplicamos filtros, agrupaciones y condiciones sobre el dataset `samples.tpch` para analizar pedidos y productos.

### Ejercicio 2.1 — Filtrar pedidos


```python
%sql
-- Mostrar pedidos con estado 'F' y monto total mayor a 10000
SELECT 
  o_orderkey AS clave_pedido,
  o_custkey AS cliente,
  o_orderstatus AS estado,
  o_totalprice AS monto_total
FROM samples.tpch.orders
WHERE o_orderstatus = 'F' AND o_totalprice > 10000
ORDER BY o_totalprice DESC;
```

### Ejercicio 2.2 — Agrupar pedidos por estado

Contamos cuántos pedidos hay por cada estado y sumamos el monto total de cada grupo.

```python
%sql
SELECT 
  o_orderstatus AS estado,
  COUNT(*) AS cantidad_pedidos,
  SUM(o_totalprice) AS monto_total
FROM samples.tpch.orders
GROUP BY o_orderstatus;
```

### Ejercicio 2.3 — Filtrar grupos con HAVING

Mostramos solo los estados con más de 10000 pedidos.

```python
%sql
SELECT 
  o_orderstatus AS estado,
  COUNT(*) AS cantidad_pedidos,
  SUM(o_totalprice) AS monto_total
FROM samples.tpch.orders
GROUP BY o_orderstatus
HAVING COUNT(*) > 10000;
```

**Diferencia entre WHERE y HAVING:**  
`WHERE` filtra filas **antes** de agrupar los datos, mientras que `HAVING` filtra **después** de aplicar la agrupación.


### Ejercicio 2.4 — Top 10 productos más pedidos

Usamos la tabla `lineitem` para encontrar los 10 productos con mayor cantidad total pedida.

```python
%sql
SELECT 
  l_partkey AS producto,
  SUM(l_quantity) AS cantidad_total
FROM samples.tpch.lineitem
GROUP BY l_partkey
ORDER BY cantidad_total DESC
LIMIT 10;
```

# Parte 3 - Combinando tablas con JOIN

Los clientes tienen c_nationkey que se relaciona con n_nationkey en nation . Los pedidos
se vinculan por o_custkey = c_custkey .


En esta sección unimos tablas del dataset `samples.tpch` para relacionar clientes, pedidos y países, aplicando `JOIN`, `GROUP BY` y funciones de agregación.

### Ejercicio 3.1 — Primer JOIN: clientes y sus pedidos

Unimos `customer` con `orders` para mostrar el nombre del cliente y el monto de cada pedido.

```python
%sql
-- Unir clientes con pedidos para ver nombre y monto
SELECT 
  c.c_name AS nombre_cliente,
  o.o_orderkey AS clave_pedido,
  o.o_totalprice AS monto_total
FROM samples.tpch.customer c
INNER JOIN samples.tpch.orders o
  ON c.c_custkey = o.o_custkey
LIMIT 20;
```

### Ejercicio 3.2 — Clientes por país

Unimos `customer` con `nation` para mostrar el nombre del cliente y el país.

```python
%sql
-- Unir clientes con países usando la clave foránea
SELECT 
  c.c_name AS nombre_cliente,
  n.n_name AS pais
FROM samples.tpch.customer c
INNER JOIN samples.tpch.nation n
  ON c.c_nationkey = n.n_nationkey;
```

### Ejercicio 3.3 — Total gastado por país

Combinamos `customer`, `orders` y `nation` para calcular cuánto gastaron en total los clientes de cada país.

```python
%sql
-- Calcular gasto total por país
SELECT 
  n.n_name AS pais,
  SUM(o.o_totalprice) AS gasto_total
FROM samples.tpch.customer c
INNER JOIN samples.tpch.orders o
  ON c.c_custkey = o.o_custkey
INNER JOIN samples.tpch.nation n
  ON c.c_nationkey = n.n_nationkey
GROUP BY n.n_name
ORDER BY gasto_total DESC;
```

### Ejercicio 3.4 — Desafío final


Encontramos el cliente que más gastó en total, mostrando su nombre, país y monto.


```python
%sql
-- Cliente que más gastó en total
SELECT 
  c.c_name AS nombre_cliente,
  n.n_name AS pais,
  SUM(o.o_totalprice) AS gasto_total
FROM samples.tpch.customer c
INNER JOIN samples.tpch.orders o
  ON c.c_custkey = o.o_custkey
INNER JOIN samples.tpch.nation n
  ON c.c_nationkey = n.n_nationkey
GROUP BY c.c_name, n.n_name
ORDER BY gasto_total DESC
LIMIT 1;
```
